```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python

df = pd.read_csv("tips.csv")

print("Dataset loaded successfully!")
```

## checking missing datas

```python

print("Missing Values:")
print(df.isnull().sum())
```

```python

print("Duplicate Rows:", df.duplicated().sum())
```

```python

df.describe()
```

```python

print("Gender:", df['sex'].unique())
print("Smoker:", df['smoker'].unique())
print("Days:", df['day'].unique())
print("Meal Time:", df['time'].unique())
```

## data visualization

```python

df.hist(figsize=(10, 6), bins=15, edgecolor='black')

plt.suptitle("Distribution of Restaurant Tips Dataset")
plt.tight_layout()
plt.show()
```

```python

plt.figure(figsize=(8, 5))

sns.scatterplot(data=df, x='total_bill', y='tip')

plt.title("Total Bill vs Tip")
plt.xlabel("Total Bill")
plt.ylabel("Tip Amount")
plt.show()
```

```python

plt.figure(figsize=(7, 5))

sns.barplot(data=df, x='sex', y='tip')

plt.title("Average Tip by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Tip")
plt.show()
```

## analyze

```python

plt.figure(figsize=(8, 5))

sns.barplot(data=df, x='day', y='tip')

plt.title("Average Tip by Day")
plt.xlabel("Day")
plt.ylabel("Average Tip")
plt.show()
```

```python

plt.figure(figsize=(7, 5))

sns.barplot(data=df, x='time', y='tip')

plt.title("Average Tip by Meal Time")
plt.xlabel("Meal Time")
plt.ylabel("Average Tip")
plt.show()
```

```python

plt.figure(figsize=(7, 5))

sns.barplot(data=df, x='smoker', y='tip')

plt.title("Average Tip by Smoking Status")
plt.xlabel("Smoking Status")
plt.ylabel("Average Tip")
plt.show()
```

```python

plt.figure(figsize=(7, 5))

sns.heatmap(df.select_dtypes(include='number').corr(),
            annot=True,
            cmap='coolwarm',
            fmt='.2f')

plt.title("Correlation Heatmap")
plt.show()
```

```python

print("Average Total Bill:", df['total_bill'].mean())
print("Average Tip:", df['tip'].mean())
print("Highest Tip:", df['tip'].max())
print("Lowest Tip:", df['tip'].min())
print("Average Party Size:", df['size'].mean())
```

## insights

```python

print("========== KEY FINDINGS ==========")

print("\n1. Average Tip by Day:")
print(df.groupby('day')['tip'].mean().sort_values(ascending=False))

print("\n2. Average Tip by Gender:")
print(df.groupby('sex')['tip'].mean())

print("\n3. Average Tip by Meal Time:")
print(df.groupby('time')['tip'].mean())

print("\n4. Average Tip by Smoking Status:")
print(df.groupby('smoker')['tip'].mean())

print("\n5. Correlation between Total Bill and Tip:")
print(df['total_bill'].corr(df['tip']))
```

```python

df.to_csv("cleaned_tips_dataset.csv", index=False)

print("Dataset saved successfully!")
```
